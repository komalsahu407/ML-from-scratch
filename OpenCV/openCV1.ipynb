```python
import cv2

# Load Car Cascade
car_cascade = cv2.CascadeClassifier(r"C:\Users\dell\Desktop\machine learning\opencv2\cars.xml")

# Load Traffic Video
cap = cv2.VideoCapture(r"C:\Users\dell\Desktop\machine learning\opencv2\Car Video 2026-07-13 at 10.45.55 PM.mp4")

# Check if XML loaded
if car_cascade.empty():
    print("Error: Car XML file not loaded.")
    exit()

# Check if video opened
if not cap.isOpened():
    print("Error: Video not found.")
    exit()

while True:
    ret, frame = cap.read()

    if not ret:
        print("Video End")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect Cars
    cars = car_cascade.detectMultiScale(gray, 1.1, 2)

    # Draw Rectangle
    for (x, y, w, h) in cars:
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv2.putText(frame, "Car", (x, y-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6,
                    (0, 255, 0), 2)

    cv2.imshow("Car Detection", frame)

    if cv2.waitKey(30) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
