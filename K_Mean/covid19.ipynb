```python
df.drop(columns='Province/State', inplace=True)
```

```python
#Data convert
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
```

```python
df.drop(columns='Date', inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Country/Region'] = le.fit_transform(df['Country/Region'])

df['WHO Region'] = le.fit_transform(df['WHO Region'])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = scaler.fit_transform(df)
```

```python
#features selection

X = df[['Country/Region',
        'Lat',
        'Long',
        'Confirmed',
        'Deaths',
        'Recovered',
        'Active',
        'WHO Region',
        'Year',
        'Month',
        'Day']]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled.shape
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
```

```python
df['Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
df.head()
```

```python
df['Cluster'].value_counts()
```

```python
df.groupby('Cluster').mean(numeric_only=True)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df['Confirmed'],
            df['Deaths'],
            c=df['Cluster'])

plt.xlabel("Confirmed Cases")
plt.ylabel("Deaths")
plt.title("COVID-19 Clusters")

plt.show()
```

```python
kmeans.cluster_centers_
```

```python
df.groupby('Cluster')[['Confirmed','Deaths','Recovered','Active']].mean()
```

```python
df['Cluster'].value_counts()
```

```python
plt.figure(figsize=(6,4))

df['Cluster'].value_counts().plot(kind='bar')

plt.xlabel("Cluster")

plt.ylabel("Number of Records")

plt.title("Cluster Distribution")

plt.show()
```
