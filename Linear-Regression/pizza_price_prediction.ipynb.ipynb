```python
df["price_rupiah"]=df["price_rupiah"].str.replace("Rp","")
```

```python
df["price_rupiah"]=df["price_rupiah"].str.replace(",","")
```

```python
df["price_rupiah"]=df["price_rupiah"].astype(int)
```

```python
#Diameter
df["diameter"]=df["diameter"].str.replace(" inch","")
df["diameter"]=df["diameter"].astype(float)
```

```python
#Label encoding

from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()

df["company"]=encoder.fit_transform(df["company"])
df["topping"]=encoder.fit_transform(df["topping"])
df["variant"]=encoder.fit_transform(df["variant"])
df["size"]=encoder.fit_transform(df["size"])
df["extra_sauce"]=encoder.fit_transform(df["extra_sauce"])
df["extra_cheese"]=encoder.fit_transform(df["extra_cheese"])
df["extra_mushrooms"]=encoder.fit_transform(df["extra_mushrooms"])
```

```python
df
```

```python
X=df.drop("price_rupiah", axis=1)

Y=df["price_rupiah"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import *
from sklearn.metrics import r2_score
```

```python
X_train, X_test, Y_train, Y_test =train_test_split(X,Y, test_size=0.3,random_state=42) 
Y_test
```

```python
#Multiple Linear Regression
from sklearn.linear_model import LinearRegression

regression = LinearRegression()
regression.fit(X_train, Y_train)
```

```python
X_test_predict = regression.predict(X_test)
X_test_predict
```

```python
from sklearn.metrics import r2_score
score=r2_score(Y_test, X_test_predict)
print(score)
```

```python
ridge_regression=Ridge(alpha=0.8)
ridge_regression.fit(X_train,Y_train)
```

```python
X_test_predict_ridge=ridge_regression.predict(X_test)
r2_score(Y_test,X_test_predict_ridge)
```

```python
lasso_regression=Lasso(alpha=0.8)
lasso_regression.fit(X_train,Y_train)
```

```python
X_test_predict_lasso=lasso_regression.predict(X_test)
r2_score(Y_test,X_test_predict_lasso)
```
